## Exploratory Data Analysis of Thailand rice prices

Dataset link: https://www.macrotrends.net/3757/global-thailand-rice-prices

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
from numpy import where
from numpy import astype
from numpy import float64
```

```python
prices = pd.read_csv("global_thai_rice_prices.csv")
```

```python
prices[['m', 'd', 'y']] = prices["Date"].str.split('/', expand=True)
prices["Date"] = prices['y'] + '-' + prices['m']
prices = prices.drop(columns=['m', 'd', 'y'])
prices.rename(columns={'Date': 'date', 'Value': 'prices'}, inplace=True)
prices.head()
```

### Correlation matrix

```python
ph_dataset = pd.read_csv("dataset.csv")
```

```python
ph_dataset = ph_dataset.merge(prices, on="date", how='inner')
```

```python
corr_matrix = ph_dataset.drop(columns=['date']).corr()

plt.figure(figsize=(6,  4))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap')
```
